```python
import pandas as pd
import numpy as np
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.action_chains import ActionChains
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
import pandas as pd
import time
```

```python
# 웹드라이버 설정
driver = webdriver.Chrome()  # chromedriver 경로를 PATH에 추가하거나 직접 지정 필요
#  lifestyle-entertainment, internet-of-things, support, business-intelligence, app-families, website-app-building, human-resources, artificial-intelligence, it-operations, communication, commerce, marketing
url = 'https://zapier.com/apps/categories/website-app-building'

driver.get(url)
driver.maximize_window()
```

```python
# 데이터 저장용 리스트 초기화
app_names = []
app_infos = []
urls = []

try:
    # 1) 첫 번째 섹션 구조 (section.css-h5v1hx-CategoryAppTable__wrapper)
    sections_1 = driver.find_elements(By.CSS_SELECTOR, "section.css-h5v1hx-CategoryAppTable__wrapper")
    for section in sections_1:
        lis = section.find_elements(By.CSS_SELECTOR, "li.css-1qwpe4d")
        for li in lis:
            try:
                name_elem = li.find_element(By.CSS_SELECTOR, ".css-x7hjk2")
                info_elem = li.find_element(By.CSS_SELECTOR, ".css-1qmcnjr-CategoryAppCard__description")
                a_tag = li.find_element(By.CSS_SELECTOR, "a.css-1ab3eul-CategoryAppCard")

                name = name_elem.text
                info = info_elem.text
                href = a_tag.get_attribute("href")

                app_names.append(name)
                app_infos.append(info)
                urls.append(href)
            except Exception as e:
                print(f"[섹션1] 데이터 추출 중 오류: {e}")

    # 2) 두 번째 섹션 구조 (section.css-18gh1cr-CategoryAppTable__wrapper)
    sections_2 = driver.find_elements(By.CSS_SELECTOR, "section.css-18gh1cr-CategoryAppTable__wrapper")
    for section in sections_2:
        lis = section.find_elements(By.CSS_SELECTOR, "li.css-1qwpe4d")
        for li in lis:
            try:
                name_elem = li.find_element(By.CSS_SELECTOR, ".css-x7hjk2")
                info_elem = li.find_element(By.CSS_SELECTOR, ".css-1qmcnjr-CategoryAppCard__description")
                a_tag = li.find_element(By.CSS_SELECTOR, "a.css-1ohnwz2-CategoryAppCard")

                name = name_elem.text
                info = info_elem.text
                href = a_tag.get_attribute("href")

                app_names.append(name)
                app_infos.append(info)
                urls.append(href)
            except Exception as e:
                print(f"[섹션2] 데이터 추출 중 오류: {e}")

    # 최종 DataFrame 생성 및 CSV 저장
    data = pd.DataFrame({
        "AppName": app_names,
        "AppInfo": app_infos,
        "Url": urls
    })

    # lifestyle-entertainment, IoT, Support, BizIntel, AppFamilies, Website-AppBuilding, HumanResources, ArtificialIntelligence, Content-Files, Productivity, ITOperations, Communication, Commerce, Marketing, Sales-CRM
    data.to_csv("D:/code/kka_1/data/Zapier_Sales-CRM_apps.csv", index=False, encoding="utf-8")
    print("Data saved!")

except Exception as e:
    print(f"An error occurred during execution: {e}")
```
