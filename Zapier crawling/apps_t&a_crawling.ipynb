```python
import pandas as pd
import numpy as np
from selenium import webdriver
from selenium.webdriver.common.by import By
from selenium.webdriver.common.keys import Keys
from selenium.webdriver.support.ui import WebDriverWait
from selenium.webdriver.support import expected_conditions as EC
from selenium.webdriver.common.action_chains import ActionChains
from selenium.common.exceptions import NoSuchElementException, ElementNotInteractableException 
import time
```

```python
# 2. CSV 파일 읽기
# lifestyle-entertainment, IoT, Support, BizIntel, AppFamilies, Website-AppBuilding, HumanResources, ArtificialIntelligence, Content-Files, Productivity, ITOperations, Communication, Commerce, Marketing, Sales-CRM
data = pd.read_excel("D:/code/kka_1/data/Zapier_AppFamilies_apps.xlsx")
print(data.shape)
data.head()
```

```python
# 3. url 리스트 추출
url_list = data["Url"].tolist()
print(len(url_list), url_list[0])
app_list = data["AppName"].tolist()
print(len(app_list), app_list[0])
```

```python
print(len(url_list))
print(url_list[0])
```

```python
urls = url_list
apps = app_list
driver = webdriver.Chrome()

for idx, url in enumerate(urls):
    print(f"\n[{idx+1}/{len(urls)}] 현재 URL: {url}")
    driver.get(url)
    driver.maximize_window()
    # WebDriverWait 객체 생성
    wait = WebDriverWait(driver, np.random.uniform(1,3))
    # container = wait.until(EC.presence_of_element_located((By.XPATH, '//*[@id="main"]/div/div[3]/div[4]/div/section/div[2]')))
#      container = wait.until(EC.presence_of_element_located((By.CSS_SELECTOR, 'div.css-ywpy44')))
    '//*[@id="main"]/div/div[3]/div[4]/div, //*[@id="main"]/div/div[3]/div[4]/div/section/div[2]'
    # 버튼이 더 이상 존재하지 않을 때까지 클릭
    while True:
        try:
            # 버튼이 존재하고 클릭 가능할 때까지 대기
    #         button = wait.until(EC.element_to_be_clickable((By.XPATH, button_xpath)))button class="_base-button_rr5dn_1
            # //*[@id="zap-template-list"]/div[2]/div/button
    #         button = wait.until(EC.element_to_be_clickable((By.XPATH, '//button[@class, "_base-button_rr5dn_1"')))
            button = wait.until(EC.element_to_be_clickable((By.XPATH, '//*[@id="main"]/div/div[3]/div[5]/div/section/div[3]/button')))
    #         button = wait.until(EC.element_to_be_clickable((By.CSS_SELECTOR, 'button._base-button_rr5dn_1')))
            # 버튼이 화면 내에 위치하도록 스크롤 조정
    #         driver.execute_script("arguments[0].scrollIntoView();", button)
            driver.execute_script("window.scrollTo(0, document.body.scrollHeight);")
            time.sleep(1)  # 스크롤 반영 대기
            # JavaScript를 사용하여 클릭
            driver.execute_script("arguments[0].click();", button)
            # 클릭 후 페이지가 반응할 시간을 부여
            time.sleep(np.random.uniform(2,3))            
        except Exception as e:
            # print("더 이상 클릭할 버튼 없음:", e)
            break  # 버튼이 없으면 루프 종료
    # 모든 앱 액션 div 찾기
    app_actions = driver.find_elements(By.XPATH, "//div[contains(@class, 'css-1ybx9px')]")
    print(len(app_actions))
    # 데이터 저장 리스트
    actioin_list = []
    for action in app_actions:
        try:
    #        category = action.find_element(By.CLASS_NAME, "_tootip-wrapper_8x43p_1").text
            category = action.find_element(By.CSS_SELECTOR, "div._tooltip-wrapper_8x43p_1 span").text  # TnA Category : <div class="_tooltip-wrapper_8x43p_1">
        except:
            category = None

        try:
    #         do = action.find_element(By.XPATH, "//span[contains(@class, 'css-1kefmdn')]").text
            do = action.find_element(By.CSS_SELECTOR, "span.css-1kefmdn").text # TnA Do  : <span class="css-1kefmdn">
        except:
            do = None

        try:
            name = action.find_element(By.CLASS_NAME, "app-action__title.css-pgz5n6").text # TnA Name : <span class="app-action__title css-pgz5n6">
    #          name = action.find_element(By.XPATH, "//span[contains(@class, 'app-action__title css-pgz5n6')]").text
        except:
            name = None

        try:
            description = action.find_element(By.CLASS_NAME, "css-1nnqqud").text # TnA Description : <p class="css-1nnqqud">
        except:
            description = None
        actioin_list.append([category, do, name, description])
    df = pd.DataFrame(actioin_list, columns=["Category", "Do", "Name", "Description"])
    df.to_csv("D:/code/kka_1/data/Apps_TnA/AppFamilies/{}_TrigActList.csv".format(apps[idx]), encoding='UTF-8', index=False)
    time.sleep(np.random.uniform(2,3))
    print(df.shape)
```
